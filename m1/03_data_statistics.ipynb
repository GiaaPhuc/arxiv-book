# Thống kê corpus — số mẫu

**Lưu ý:** Notebook này đọc file CSV mẫu trong `stats/`, do D chuẩn bị từ corpus giả trong repo mã (`corpus_stats.ipynb`). Đây **không phải** kết quả crawl 25.000 ID. Sau Phase 2, thay file trong `stats/` bằng số thật từ `export_stats` / pipeline.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / "stats").is_dir():
    ROOT = next(p for p in Path.cwd().parents if (p / "stats" / "sample_papers.csv").is_file())

papers_df = pd.read_csv(ROOT / "stats" / "sample_papers.csv")
versions_df = pd.read_csv(ROOT / "stats" / "sample_versions.csv")
papers_df, versions_df
```

```python
n_papers = len(papers_df)
n_with_tex = int(papers_df["has_tex"].sum())
success_rate = n_with_tex / n_papers if n_papers else 0.0
avg_refs = float(papers_df["n_refs"].mean()) if n_papers else 0.0
avg_kept = float(versions_df["kept_bytes"].mean()) if len(versions_df) else 0.0
avg_raw = float(versions_df["raw_uncompressed_bytes"].mean()) if len(versions_df) else 0.0

summary = pd.DataFrame(
    {
        "metric": [
            "papers (sample)",
            "share with TeX (sample)",
            "avg references per paper (sample)",
            "avg raw_uncompressed_bytes (sample)",
            "avg kept_bytes (sample)",
        ],
        "value": [
            n_papers,
            f"{success_rate:.0%}",
            f"{avg_refs:.2f}",
            f"{avg_raw:.0f}",
            f"{avg_kept:.0f}",
        ],
    }
)
summary
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
row = versions_df.iloc[0]
ax.bar(
    ["raw (uncompressed)", "kept (.tex+.bib)"],
    [row["raw_uncompressed_bytes"], row["kept_bytes"]],
    color=["#4C72B0", "#55A868"],
)
ax.set_ylabel("bytes (sample)")
ax.set_title(f"Kích thước phiên bản mẫu {row['arxiv_id']} v{row['version']}")
plt.tight_layout()
plt.show()
```
